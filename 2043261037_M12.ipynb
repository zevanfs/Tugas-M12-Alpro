```python
#Nomor 1
class karyawan:
    def __init__(self, nama, gaji):
        self.nama = nama
        self.gaji = int(gaji)

daftar_karyawan = [
    karyawan("Siti", "5000000"),
    karyawan("Andi", "6789000"),
    karyawan("Budi", "7654000"),
    karyawan("Caca", "8910000"),
    karyawan("Dita", "9101000"),
]

for k in daftar_karyawan:
    print(f"Karyawan {k.nama} memiliki gaji Rp{k.gaji:,}")
```

```python
#Nomor 2
class karyawan:
    def __init__(self, nama, gaji):
        self.nama = nama
        self.gaji = int(gaji)
        self.energi = 100
        
    def lembur(self):
        print(f"{self.nama} sedang lembur!")
        self.energi -= 30
        self.gaji += 200000  

daftar_karyawan = [
    karyawan("Siti", "5000000"),
    karyawan("Andi", "6789000"),
    karyawan("Budi", "7650000"),
    karyawan("Caca", "8910000"),
    karyawan("Dita", "9100000"),
]

for k in daftar_karyawan:
    k.lembur()
    print(f"Sisa energi  : {k.energi}%")
    print(f"Gaji terbaru : Rp{k.gaji:,}\n")
```

```python
#Nomor 3
class barang:
    def __init__(self, nama, harga):
        self.nama = nama
        self.harga = int(harga)

barang1 = barang("Baju", "100000")
print(f"Harga {barang1.nama} adalah Rp{barang1.harga:,}")

barang1.harga = 75000
print(f"Harga {barang1.nama} setelah diskon adalah Rp{barang1.harga:,}")
```

```python
#nomor 4
class RekeningBank:
    def __init__(self, nama_pemilik, saldo = 0):
        self.nama_pemilik = nama_pemilik
        self.saldo = int(saldo)

    def menabung(self, nominal):
        self.saldo += nominal 
        print(f"Berhasil menabung Rp{nominal:,}. Saldo sekarang: Rp{self.saldo:,}.")

    def tarik_tunai(self, nominal):
        if self.saldo >= nominal:
            self.saldo -= nominal
            print(f"Berhasil menarik Rp{nominal:,}. Saldo sekarang: Rp{self.saldo:,}.")
        else:
            print(f"Maaf, saldo tidak mencukupi.")

rekening = RekeningBank("Zeva", 150000)
rekening.tarik_tunai(50000)
rekening.tarik_tunai(200000)
rekening.menabung(300000)     
```
